```python
import torch
import torch.nn as nn

class SpaNet_Attention(nn.Module):
    """
    Implements a Spatial Attention Module, often inspired by CBAM's spatial block.
    It pools across the channel dimension and applies a convolution to create
    a spatial attention map.
    """
    def __init__(self, kernel_size=7):
        """
        :param kernel_size: The kernel size for the convolutional layer.
                            Commonly 7 or 3.
        """
        super(SpaNet_Attention, self).__init__()

        # The output channel is 1 because we are creating a single attention map
        # to multiply across all channels of the input feature map.
        self.conv = nn.Conv2d(
            in_channels=2, # GAP and GMP are concatenated, so 2 channels
            out_channels=1,
            kernel_size=kernel_size,
            stride=1,
            padding=kernel_size // 2, # Maintain spatial dimensions
            bias=False
        )
        self.sigmoid = nn.Sigmoid()

        print(f"Initialized SpaNet_Attention with kernel_size={kernel_size}")

    #

    def forward(self, x):
        # x shape: (B, C, H, W)

        # 1. Global Max Pooling across channel dimension (C)
        # Output shape: (B, 1, H, W)
        max_out, _ = torch.max(x, dim=1, keepdim=True)

        # 2. Global Average Pooling across channel dimension (C)
        # Output shape: (B, 1, H, W)
        avg_out = torch.mean(x, dim=1, keepdim=True)

        # 3. Concatenate the pooled features along the channel dimension (dim=1)
        # Concatenated shape: (B, 2, H, W)
        combined = torch.cat([max_out, avg_out], dim=1)

        # 4. Apply a 2D convolution to generate the spatial attention map
        # Attention map shape: (B, 1, H, W)
        attention_map = self.sigmoid(self.conv(combined))

        # 5. Apply attention to the input feature map
        # The attention map is broadcasted across the channel dimension (C)
        out = x * attention_map

        return out

print("✅ SpaNet_Attention class defined (using standard Spatial Attention mechanism).")

# Example usage (Optional):
# input_features = torch.randn(4, 64, 32, 32) # Batch, Channels, Height, Width
# spa_attn = SpaNet_Attention()
# output = spa_attn(input_features)
# print(f"Input shape: {input_features.shape}, Output shape: {output.shape}")
```

```python
import torch
import torch.nn as nn
# Assuming the necessary imports from ultralytics are available in the execution environment
try:
    from ultralytics.nn.modules import Conv, C2f
except ImportError:
    # Define a minimal Conv class for execution context if ultralytics is not available
    class Conv(nn.Module):
        def __init__(self, c1, c2, k=1, s=1, p=None, g=1, act=True):
            super().__init__()
            self.conv = nn.Conv2d(c1, c2, k, s, k // 2 if p is None else p, groups=g, bias=False)
            self.bn = nn.BatchNorm2d(c2)
            self.act = nn.ReLU() if act is True else (act if isinstance(act, nn.Module) else nn.Identity())

        def forward(self, x):
            return self.act(self.bn(self.conv(x)))

    # Define a dummy C2f for type annotation if needed, but we redefine it anyway.
    class C2f(nn.Module):
        pass # Placeholder

# Define SpaNet_Attention (Spatial Attention)
class SpaNet_Attention(nn.Module):
    """
    Implements a standard Spatial Attention Module (similar to CBAM's spatial block).
    It pools across the channel dimension and applies a convolution to create
    a spatial attention map.
    """
    def __init__(self, kernel_size=7):
        super(SpaNet_Attention, self).__init__()

        # Convolves the 2 channels (from max and avg pooling) into 1 attention map channel.
        self.conv = nn.Conv2d(
            in_channels=2,
            out_channels=1,
            kernel_size=kernel_size,
            stride=1,
            padding=kernel_size // 2, # Maintain spatial dimensions
            bias=False
        )
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        # x shape: (B, C, H, W)

        # Global Max Pooling across channel dimension (C) -> (B, 1, H, W)
        max_out, _ = torch.max(x, dim=1, keepdim=True)

        # Global Average Pooling across channel dimension (C) -> (B, 1, H, W)
        avg_out = torch.mean(x, dim=1, keepdim=True)

        # Concatenate the pooled features along the channel dimension (dim=1) -> (B, 2, H, W)
        combined = torch.cat([max_out, avg_out], dim=1)

        # Apply convolution and Sigmoid to generate the spatial attention map -> (B, 1, H, W)
        attention_map = self.sigmoid(self.conv(combined))

        # Apply attention to the input feature map (broadcasting 1 channel across C channels)
        out = x * attention_map

        return out

print("✅ SpaNet_Attention class defined (Spatial Attention).")

# Refactored Module: C2f_Spa

# The `C2f_ASR` class is refactored to **`C2f_Spa`**, replacing `ASR_Attention` with `SpaNet_Attention`. The input channels for `SpaNet_Attention` are not needed, as the spatial module pools across channels dynamically.

class C2f_Spa(nn.Module):
    """C2f block with SpaNet-style Spatial Attention module added after the bottleneck."""
    def __init__(self, c1, c2, n=1, shortcut=False, g=1, e=0.5):
        super().__init__()
        self.c = int(c2 * e)  # hidden channels
        self.cv1 = Conv(c1, 2 * self.c, 1, 1)
        # Note: shortcut is not used in the YOLOv8 C2f implementation provided here,
        # but kept for compatibility with the C2f signature.
        self.cv2 = Conv((2 + n) * self.c, c2, 1)
        self.m = nn.ModuleList(
            nn.Sequential(
                Conv(self.c, self.c, 3, 1, g=g),
                Conv(self.c, self.c, 3, 1, g=g)
            ) for _ in range(n)
        )
        # ✅ THE SpaNET ATTENTION INJECTION ✅
        # Spatial attention (SpaNet/CBAM style) does not require 'c2' in its __init__.
        self.att = SpaNet_Attention(kernel_size=7) # Use the Spatial Attention module

    def forward(self, x):
        # 1. YOLOv8 C2f Logic (Splitting and Sequential Blocks)
        y = list(self.cv1(x).chunk(2, 1))
        y.extend(m(y[-1]) for m in self.m)
        out = self.cv2(torch.cat(y, 1))

        # 2. Apply SpaNet Attention on the output features
        return self.att(out)

print("✅ C2f_Spa block defined, integrating SpaNet_Attention.")
```

```python
import torch
import torch.nn as nn
from ultralytics.nn.modules import Conv, C2f # Required for model inspection
from ultralytics import YOLO

# Note: ASR_Attention and C2f_ASR definitions are removed/not needed here as
# SpaNet_Attention and C2f_Spa are defined in a prior cell (A0a51LwOlATr) and will be used.

def create_spanet_yolo_model(model_path='yolov8n.pt'):
    """
    Loads a standard YOLO model and replaces C2f blocks
    with C2f_Spa blocks in the backbone/head.
    """
    print(f"🏗 Loading base model: {model_path}...")
    model = YOLO(model_path)

    print("🔁 Injecting SpaNet Spatial Attention Modules into architecture...")

    nn_model = model.model

    for name, module in nn_model.named_modules():
        if isinstance(module, C2f):
            parent_name = name.rsplit('.', 1)[0] if '.' in name else ''
            child_name = name.rsplit('.', 1)[-1]

            if parent_name:
                parent = nn_model.get_submodule(parent_name)
            else:
                parent = nn_model

            # Extract parameters from existing C2f
            c1 = module.cv1.conv.in_channels
            c2 = module.cv2.conv.out_channels
            n = len(module.m)

            # Create new SpaNet C2f
            new_module = C2f_Spa(c1, c2, n=n) # Using C2f_Spa defined earlier

            # Replace
            setattr(parent, child_name, new_module)

    print("✅ Model transformation complete. SpaNet Attention active.")
    return model

# Usage: Create the SpaNet-enhanced model
model_spanet = create_spanet_yolo_model('yolov8n.pt')

print("✅ STEP 3 COMPLETED: SpaNet Attention Architecture ready.")
```

```python
import os
from pathlib import Path
from PIL import Image # Pillow for image size
import shutil # Import shutil for file operations
from sklearn.model_selection import train_test_split

print("🛠 Starting dataset conversion to YOLO format...")

source_dataset_root = '/content/datasets/figshare_mri' # Updated source path
target_dataset_root = '/content/datasets/figshare_mri_yolo' # Updated target path

# Class mapping for YOLO labels - UPDATED CLASS NAMES
class_names = ['glioma_tumor', 'meningioma_tumor', 'no_tumor', 'pituitary_tumor']
class_to_id = {name: i for i, name in enumerate(class_names)}

# Create target directories
for split in ['train', 'val', 'test']:
    Path(f'{target_dataset_root}/{split}/images').mkdir(parents=True, exist_ok=True)
    Path(f'{target_dataset_root}/{split}/labels').mkdir(parents=True, exist_ok=True)

# Process Training split to create train and validation sets
training_source_path = Path(source_dataset_root) / 'Training'
if training_source_path.exists():
    print(f"Processing Training split for train/val...")
    all_training_images = []
    for class_folder in training_source_path.iterdir():
        if class_folder.is_dir():
            class_name = class_folder.name.lower()
            if class_to_id.get(class_name) is not None:
                all_training_images.extend(list(class_folder.glob('*.jpg')))

    # Split training images into train and validation
    train_images, val_images = train_test_split(all_training_images, test_size=0.2, random_state=42)

    # Copy images and create labels for the training set
    for img_file in train_images:
        class_name = img_file.parent.name.lower()
        class_id = class_to_id[class_name]
        shutil.copy(img_file, Path(target_dataset_root) / 'train' / 'images' / img_file.name)
        label_filename = img_file.stem + '.txt'
        with open(Path(target_dataset_root) / 'train' / 'labels' / label_filename, 'w') as f:
            f.write(f"{class_id} 0.5 0.5 1.0 1.0\n")

    # Copy images and create labels for the validation set
    for img_file in val_images:
        class_name = img_file.parent.name.lower()
        class_id = class_to_id[class_name]
        shutil.copy(img_file, Path(target_dataset_root) / 'val' / 'images' / img_file.name)
        label_filename = img_file.stem + '.txt'
        with open(Path(target_dataset_root) / 'val' / 'labels' / label_filename, 'w') as f:
            f.write(f"{class_id} 0.5 0.5 1.0 1.0\n")
else:
    print(f"⚠️  Warning: Original training folder '{training_source_path}' not found. No train/val split created.")

# Process Testing split for the test set
testing_source_path = Path(source_dataset_root) / 'Testing'
if testing_source_path.exists():
    print(f"Processing Testing split for test...")
    for class_folder in testing_source_path.iterdir():
        if class_folder.is_dir():
            class_name = class_folder.name.lower()
            class_id = class_to_id.get(class_name)

            if class_id is None:
                print(f"  Skipping unknown class folder: {class_name}")
                continue

            for img_file in class_folder.glob('*.jpg'):
                shutil.copy(img_file, Path(target_dataset_root) / 'test' / 'images' / img_file.name)
                label_filename = img_file.stem + '.txt'
                with open(Path(target_dataset_root) / 'test' / 'labels' / label_filename, 'w') as f:
                    f.write(f"{class_id} 0.5 0.5 1.0 1.0\n")
else:
    print(f"⚠️  Warning: Original testing folder '{testing_source_path}' not found. No test split created.")

print(f"✅ Dataset conversion complete. YOLO formatted data is at {target_dataset_root}")
```
